```python
import gymnasium as gym
import numpy as np
import random
```

### Hyperparameter Configuration
Defined the number of training episodes, learning rate (α), discount factor (γ), and exploration rate (ε) for the SARSA agent.

```python
episodes = 500
alpha = 0.5
gamma = 0.99
epsilon = 0.1
```

## Environment Setup

```python
env = gym.make('CliffWalking-v1')
```

## Initialized Q-table With Zeros

```python
Q = np.zeros((48,4))
```

## ε-greedy policy

```python
def greedy_epsilon(state):
    if random.random() < epsilon:
        return env.action_space.sample() #EXPLORE
    else:
        return np.argmax(Q[state]) # EXPLOIT
        
```

## Q-Learning Implementation

```python
for episode in range(episodes):
    env = gym.make('CliffWalking-v1')
    done = False
    total_reward = 0
    epsiode_length = 0

    state,_ = env.reset()
    
    while not done:
        action = greedy_epsilon(state)
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        Q[state, action] += alpha * (reward + gamma * np.max(Q[next_state]) -  Q[state,action])
        state = next_state

        total_reward += reward
        epsiode_length += 1
    print(f"For episode {episode+1}/{episodes}: reward={total_reward} & episode_length={epsiode_length}")
    env.close()
    
```

## Q-Learning Evaluation

```python
env = gym.make('CliffWalking-v1', render_mode = 'human')
done = False
total_reward = 0
episode_length = 0
state, _ = env.reset()

while not done:
    action = np.argmax(Q[state])
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    total_reward += reward
    episode_length += 1
print(f"total reward = {total_reward} & episode len = {episode_length}")
env.close()
```
