```python
import gymnasium as gym
import numpy as np
import random
```

### Hyperparameter Configuration
Defined the number of training episodes, learning rate (α), discount factor (γ), and exploration rate (ε) for the SARSA agent.

```python
episodes = 500
alpha = 0.5
gamma = 0.99
epsilon = 0.1
```

## Environment Setup

```python
env = gym.make('CliffWalking-v1')
```

## Initialized Q-table With Zeros

```python
Q = np.zeros((48,4))
```

## ε-greedy policy

```python
def greedy_epsilon(state):
    if random.random() < epsilon:
        return env.action_space.sample() #EXPLORE
    else:
        return np.argmax(Q[state]) # EXPLOIT
        
```

### SARSA Training
Trained the agent over multiple episodes using the SARSA update rule, ε-greedy action selection, and tracked the total reward and episode length.

```python
for episode in range(episodes):
    env = gym.make('CliffWalking-v1')
    
    done = False
    total_reward = 0.0
    epsiode_length = 0.0
    
    state, _ = env.reset()
    action = greedy_epsilon(state)
    
    while not done:
        next_state,reward,terminated,truncated,_ = env.step(action)
        next_action = greedy_epsilon(next_state)
        Q[state,action] += alpha * (reward + gamma * (Q[next_state,next_action] - Q[state,action])) #Q update
        
        done = terminated or truncated
        total_reward += reward
        epsiode_length += 1
        
        state = next_state
        action = next_action
    print(f"For episode {episode+1}/{episodes}: reward={total_reward} & episode_length={epsiode_length}")
    env.close()
```

### Agent Evaluation
Evaluated the learned Q-table by selecting the highest-valued action at each state and observing the agent's performance in the CliffWalking environment.

```python
# what did our agent learn?
env = gym.make('CliffWalking-v1',render_mode ='human')
state,_ = env.reset()
done = False
total_reward = 0
epsiode_length = 0 

while not done:
    action = np.argmax(Q[state])
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
    
    total_reward += reward
    epsiode_length += 1
    
print(f"total reward = {total_reward} & episode len = {epsiode_length}")
env.close()
    
```
